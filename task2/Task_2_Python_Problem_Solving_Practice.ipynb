## Q1. Find the Shortest Word in a Sentence

### 1. Observe
Input: a sentence. Output: its shortest word.

### 2. Understand
Split the sentence into words and select the word with the fewest characters.

### 3. Break
1. Split the sentence. 2. Compare word lengths. 3. Return the first word with the minimum length.

### 4. Logic
Set the first word as the current shortest word. Check each remaining word and update only when a shorter word is found.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `easy; is`


```python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    shortest = words[0]
    for word in words[1:]:
        if len(word) < len(shortest):
            shortest = word
    return shortest

print(shortest_word("Python makes coding easy"))  # Example
print(shortest_word("Learning is fun"))           # Additional test
```

## Q8. Find the Most Frequent Character

### 1. Observe
Input: a string. Output: its most frequent non-space character.

### 2. Understand
Count characters while ignoring spaces; in a tie, choose the character whose first occurrence is earliest.

### 3. Break
1. Remove spaces. 2. Count each character. 3. Scan in original order and retain the first character with the highest count.

### 4. Logic
Build a frequency dictionary, then scan the filtered string and update the answer only for a strictly greater frequency.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `a; a`


```python
def most_frequent_character(text):
    characters = [character for character in text if not character.isspace()]
    if not characters:
        return ""
    frequency = {}
    for character in characters:
        frequency[character] = frequency.get(character, 0) + 1
    result = characters[0]
    for character in characters:
        if frequency[character] > frequency[result]:
            result = character
    return result

print(most_frequent_character("banana"))       # Example
print(most_frequent_character("aabb"))         # Additional test
```

```python
def repeated_characters(text):
    characters = [character for character in text if not character.isspace()]
    frequency = {}
    for character in characters:
        frequency[character] = frequency.get(character, 0) + 1
    result = []
    for character in characters:
        if frequency[character] > 1 and character not in result:
            result.append(character)
    return result

print(repeated_characters("programming"))  # Example
print(repeated_characters("hello world"))  # Additional test
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status

### 1. Observe

### 2. Understand
A number is even when its remainder after division by 2 is zero; otherwise it is odd.

### 3. Break
1. Create an empty dictionary. 2. Check each number with modulo. 3. Store the appropriate label.

### 4. Logic
Use number % 2 == 0 to decide between 'Even' and 'Odd'.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}; {0: 'Even', -3: 'Odd', -4: 'Even'}`


```python
def even_odd_dictionary(numbers):
    result = {}
    for number in numbers:
        if number % 2 == 0:
            result[number] = "Even"
        else:
            result[number] = "Odd"
    return result

print(even_odd_dictionary([1, 2, 3, 4, 5]))  # Example
print(even_odd_dictionary([0, -3, -4]))       # Additional test
```

## Q16. Find the Most Frequent Word in a Sentence

### 1. Observe
Input: a sentence. Output: its most frequent word in lowercase.

### 2. Understand
Treat uppercase and lowercase as the same word; if counts tie, return the word whose first occurrence is earliest.

### 3. Break
1. Split and lowercase words. 2. Count each word. 3. Scan in order and keep the first word with the highest count.

### 4. Logic
Use a frequency dictionary and update the result only when a word has a strictly higher count.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `python; tea`


```python
def most_frequent_word(sentence):
    words = [word.lower() for word in sentence.split()]
    if not words:
        return ""
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    result = words[0]
    for word in words:
        if frequency[word] > frequency[result]:
            result = word
    return result

print(most_frequent_word("Python is easy and Python is powerful"))  # Example
print(most_frequent_word("Tea tea COFFEE coffee"))                  # Additional test
```

## Q17. Find All Words Appearing More Than Once

### 1. Observe
Input: a sentence. Output: repeated words once each, in first-appearance order.

### 2. Understand
Compare words case-insensitively and return only words whose frequency is greater than one.

### 3. Break
1. Convert words to lowercase. 2. Count frequencies. 3. Scan the words and append each repeated word once.

### 4. Logic
Build a frequency dictionary, then add words with frequency above one if they are not already in the result.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `['data', 'science']; ['go', 'stop']`


```python
def repeated_words(sentence):
    words = [word.lower() for word in sentence.split()]
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    result = []
    for word in words:
        if frequency[word] > 1 and word not in result:
            result.append(word)
    return result

print(repeated_words("data science uses data and science tools"))  # Example
print(repeated_words("Go go STOP stop wait"))                      # Additional test
```

### 3. Break
1. Convert words to lowercase. 2. Count each word. 3. Scan in order and return the first word with count one.

### 5. Code
Run the next cell in Jupyter Notebook.

### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `mango; two`


```python
def first_unique_word(sentence):
    words = [word.lower() for word in sentence.split()]
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    for word in words:
        if frequency[word] == 1:
            return word
    return ""

print(first_unique_word("apple banana apple mango banana orange"))  # Example
print(first_unique_word("one TWO one three"))                       # Additional test
```

## Q20. Find the Longest Word That Appears Only Once
### 1. Observe
Input: a sentence. Output: the longest word that appears exactly once.
### 2. Understand
Compare words case-insensitively; when unique words tie in length, return the first one.
### 3. Break
1. Convert words to lowercase. 2. Count word frequencies. 3. Scan in order, considering only words with frequency one, and keep the longest.
### 4. Logic
Build frequencies, then update the selected word only when a unique word is strictly longer than the current result.
### 5. Code
Run the next cell in Jupyter Notebook.
### 6. Test
The code tests the question's example and an additional test case. Expected results are noted below for comparison:
- Example and additional test: `coding; apple`

```python
def longest_unique_word(sentence):
    words = [word.lower() for word in sentence.split()]
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1

    result = ""
    for word in words:
        if frequency[word] == 1 and len(word) > len(result):
            result = word
    return result

print(longest_unique_word("Python makes coding useful Python"))  # Example
print(longest_unique_word("apple grape melon"))                   # Additional test
```
