```python
# Q35: Prediction: It prints [[1, 99], [3, 4]].
# Why: A shallow copy duplicates only the outer list; nested lists are shared.
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

```python
# Q36: Prediction: It prints [[1, 2], [3, 4]].
# Why: A deep copy recursively copies nested lists, so the mutation to b does not affect a.
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```
